## 🌟 HOW TO RUN IN BACKGROUND 24/7 WITH BROWSER CLOSED:
1. **Persistent Cloud VM**: Lightning Studios do **not** shut down when you close your browser tab! Your virtual environment, files, and running processes stay alive in the cloud.
2. **Detached Background Execution**: Run Step 4 to launch training via `nohup`. It runs completely in the background.
3. **Zero Interruption**: You can turn off your laptop, go to sleep, and check back anytime. Training runs 24/7.
4. **Continuous Cloud Sync**: Checkpoints are auto-saved to persistent studio storage (`FaceKeyDataset/checkpoints/`) AND pushed to Hugging Face Hub (`VijayTheOne/facekey-dataset-chunks`) every epoch.
5. **1-Click Auto-Resume**: If you ever restart the Studio, running the trainer immediately picks up from the latest saved checkpoint!

**🛡️ GEN2 ARCHITECTURAL ENHANCEMENTS:**
- 64-band Log-Mel spectral filterbanks.
- 5-state Emotional Expression Conditioning (`Neutral`, `Happy`, `Angry`, `Scared`, `Sad`).
- 3D Head Orientation Kinematics (`Pitch`, `Yaw`, `Roll`) with velocity loss.
- Lingual Tongue Invisibility Guard (`tongueOut`).

## 🌙 STEP 3: LAUNCH 24/7 BACKGROUND TRAINING (BROWSER-CLOSE SAFE)
> Starts the training as a detached background process. You can close your browser tab immediately!

```python
import os, subprocess

DRIVE_DIR = '/teamspace/studios/this_studio/FaceKeyDataset'
os.makedirs(f'{DRIVE_DIR}/checkpoints', exist_ok=True)
os.makedirs(f'{DRIVE_DIR}/chunks', exist_ok=True)
LOG_FILE = f'{DRIVE_DIR}/checkpoints/lightning_training.log'

# Launch detached background runner
cmd = f'''
nohup python scripts/train_speech_to_animation.py \
    --drive-dir "{DRIVE_DIR}" \
    --hf-repo "VijayTheOne/facekey-dataset-chunks" \
    --epochs 50 \
    --batch-size 64 \
    --lr 0.0001 \
    --seq-len 64 \
    --stride 16 \
    --accum-steps 2 > "{LOG_FILE}" 2>&1 &
'''
subprocess.Popen(cmd, shell=True, executable='/bin/bash')

print('[✓] FaceKey Gen2 Trainer is running in the background on Lightning.ai!')
print(f'[*] Log file: {LOG_FILE}')
print('🎉 YOU CAN NOW SAFELY CLOSE YOUR BROWSER! Training runs 24/7 in the cloud.')
```
